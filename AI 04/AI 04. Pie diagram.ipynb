```python
import matplotlib.pyplot as plt
```

```python
labels = ["Baku", "Xirdalan", "Absheron"]
percents = [110, 35, 45]
```

```python
plt.pie(percents, labels=labels)
plt.title("Pie Diagram")
plt.show()
```

```python
plt.pie(percents, labels=labels, autopct="%1.2f%%")
plt.title("Pie Diagram")
plt.show()
```

```python
explode = (0, 0.3, 0)
plt.pie(
    percents,
    labels=labels,
    autopct="%1.2f%%",
    explode=explode,
    shadow=True,
    startangle=90,
    colors=["#91c2e6", "#ffff00", "#f041ea"],
)
plt.title("Pie Diagram")
plt.show()
```
