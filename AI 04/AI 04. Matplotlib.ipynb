```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 10)
x
```

```python
y = x*2
y
```

```python
plt.plot(x, y)
plt.title('y = 2x')
plt.xlabel('x')
plt.ylabel('y')
plt.xlim(0, 10)
plt.ylim(0, 20)
plt.savefig('plot.png', dpi=300)
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes((0, 0, 1, 1)) #left, bottom, width, height
axes.plot(x, y)
```

```python
a = np.linspace(0, 10, 15)
a
```

```python
b = a**2
b
```

```python
fig = plt.figure()
axes1 = fig.add_axes((0, 0, 1, 1))
axes1.plot(a, b)
axes1.set_title("a = b^2")
axes1.set_xlabel("a")
axes1.set_ylabel("b")
axes1.set_xlim(0, 10)
axes1.set_ylim(0, 100)

axes2 = fig.add_axes((0.5, 0.5, 0.5, 0.5))    # left, bottom, width, height
axes2.plot(x, y)
axes2.set_title("y = 2x")
axes2.set_xlabel("x")
axes2.set_ylabel("y")
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot(a, b)
axes[1].plot(x, y)
plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
axes[0, 0].plot(a, b)
axes[0, 1].plot(x, y)
axes[1, 0].plot(x, np.sin(x))
axes[1, 1].plot(x, np.tan(x))
```

```python
fig = plt.figure()
axes = fig.add_axes((0, 0, 1, 1))
axes.plot(a, b, label="y=x^2", color="#189cab")
axes.plot(x, y, label="y=2x", color='green', lw=3)
axes.plot(a, 100* np.sin(a), label="b = sin(a) * 100", color='blue', lw=5)
axes.legend()
plt.show()
```
