```python
import matplotlib.pyplot as plt
```

```python
languages = ["İngilis", "Rus", "Alman", "Fransız"]

word_counts = [5000, 3500, 5600, 1500]
```

```python
plt.bar(languages, word_counts, color="red")
plt.xlabel("Language")
plt.ylabel("Word Count")
plt.title("Word Count Distribution")
plt.show()
```

```python
plt.barh(languages, word_counts, color=['blue', 'green', 'red',"orange"])
plt.xlabel("Language")
plt.ylabel("Word Count")
plt.title("Word Count Distribution")
for i , val in enumerate(word_counts):
    plt.text(i, i+0.43, str(val), ha='left', fontweight='bold', fontsize=10)
plt.show()
```
