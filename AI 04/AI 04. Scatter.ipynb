### 7. İstifadə nümunəsi

```python
import matplotlib.pyplot as plt

data = np.random.rand(10, 10)

plt.colorbar()
plt.show()
```

**Qeyd:** `cmap` parametri verilənlərin ədədi qiymətlərini rənglərə uyğunlaşdırmaq üçün istifadə olunur.


````python
x = np.random.random(100)
y = np.random.random(100)
colors = np.random.random(100)
sizes = 500 * np.random.random(100)
plt.scatter(x,y,c=colors,s=sizes, alpha=0.8, cmap="RdYlBu")
plt.show()
````


## Matplotlib `marker` dəyərləri

### 1. Əsas marker-lər

- `.` — nöqtə (point)
- `,` — piksel (pixel)
- `o` — dairə (circle)
- `v` — aşağı istiqamətli üçbucaq (triangle down)
- `^` — yuxarı istiqamətli üçbucaq (triangle up)
- `<` — sola istiqamətli üçbucaq (triangle left)
- `>` — sağa istiqamətli üçbucaq (triangle right)
- `1` — aşağı istiqamətli üçşüalı marker (tri down)
- `2` — yuxarı istiqamətli üçşüalı marker (tri up)
- `3` — sola istiqamətli üçşüalı marker (tri left)
- `4` — sağa istiqamətli üçşüalı marker (tri right)

### 2. Kvadrat və çoxbucaqlılar

- `s` — kvadrat (square)
- `p` — beşbucaq (pentagon)
- `h` — altıbucaq (hexagon 1)
- `H` — altıbucaq (hexagon 2)
- `8` — səkkizbucaq (octagon)

### 3. Ulduz və romb

- `*` — ulduz (star)
- `D` — romb (diamond)
- `d` — nazik romb (thin diamond)

### 4. Xətt tipli marker-lər

- `+` — üstəgəl (plus)
- `x` — çarpaz (x)
- `X` — doldurulmuş çarpaz (filled x)
- `P` — doldurulmuş üstəgəl (filled plus)

### 5. Digər marker-lər

- `|` — şaquli xətt (vertical line)
- `_` — üfüqi xətt (horizontal line)
- `None` — marker yoxdur
- `''` — marker yoxdur
- `' '` — marker yoxdur

### 6. İstifadə nümunəsi

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(
    x, y,
    marker="o",
    markersize=10,
    markerfacecolor="red",
    markeredgecolor="black",
    linestyle="--"
)

plt.show()
```

### 7. Marker parametrləri

| Parametr | İzah |
|---|---|
| `marker` | Markerin forması |
| `markersize` və ya `ms` | Markerin ölçüsü |
| `markerfacecolor` və ya `mfc` | Markerin daxili rəngi |
| `markeredgecolor` və ya `mec` | Markerin kənar rəngi |
| `markeredgewidth` və ya `mew` | Markerin kənar xəttinin qalınlığı |

**Qeyd:** `marker` qrafikdə məlumat nöqtələrinin formasını müəyyən edir. Xüsusilə `plt.plot()` və `plt.scatter()` funksiyalarında istifadə olunur.


````python
plt.scatter(x,y,color='red', label="Red", marker="+")
plt.scatter(y, x, color='blue', label="Blue", marker="P", s=sizes, alpha=0.8)
plt.legend()
plt.show()
````
