```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([45, 53, 65, 70, 75])
```

```python
X
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
a = model.intercept_
b = model.coef_[0]
```

```python
print(f'Tənlik: y = {a:.2f} + {b:.2f}x')
```

```python
y_pred = model.predict(X)
y_pred
```

```python
plt.scatter(X, y, color='red', label='Datas')
plt.plot(X, y, color='blue', label='Regression')
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

```python
while True:
    try:
        x_new = float(input('Enter a hours: '))
        if x_new == 0:
            print('Exit')
            break
        y_new = a + b * x_new
        print(f"Predicted mark value: {y_new:.2f}")
        plt.scatter(x_new, y_new, color='red', label='our prediction', s=100)
        plt.scatter(X, y, color='red', label='Datas')
        plt.plot(X, y, color='blue', label='Regression')
        plt.xlabel('Hours')
        plt.ylabel('Marks')
        plt.title('Simple Linear Regression')
        plt.legend()
        plt.show()
    except ValueError:
        print("Please enter a number")
```
