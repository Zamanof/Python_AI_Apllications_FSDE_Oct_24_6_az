# Hesablama qrafı - Computational graphs

$$
y = x^2 + 5x + 7
$$

1. $x^2$ hesablanır
2. $5x$ hesablanır
3. bu nətizələri cəmləyib üzərinə 7 gəlirik



```python
X = torch.tensor(2.0, requires_grad=True)
y = X**2 + 5*X + 7

X, y
```

```python
y.grad_fn
```

```python
y.backward()
print(X.grad)
```
